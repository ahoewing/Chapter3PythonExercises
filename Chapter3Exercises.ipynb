```python
seconds_per_hour = 60 * 60
print(seconds_per_hour)
```

```python
seconds_per_hour * 24
```

```python
seconds_per_day = seconds_per_hour * 24
print(seconds_per_day)
```

```python
seconds_per_day / seconds_per_hour
```

```python
seconds_per_day // seconds_per_hour
```

Yes, both answers agree. Floating-point division gave me 24.0, while integer division gave me 24. The only difference is the .0 at the end.
